# Kans / Probability

Kans = relatieve frequentie van het voorkomen van een gebeurtenis binnen een experiment (bij grote aantal keren uitvoeren van het experiment)

Compound event / samengestelde gebeurtenis = voorkomen van meerdere gebeurtenissen. 

In simpele gevallen kan je een lijst van uitkomsten maken (= sample space / universe)

Notatie: 
- Verwijzen naar de lijst: Ω
- De lijst zelf: tussen {}
- Deelverzameling van de lijst (incl. een gebeurtenis): hoofdletter
- Samengestelde gebeurtenis: (A, B, C)

Via het universum kan je de kans berekenen als => aantal voorkomens van de (samengestelde) gebeurtenis / lengte van het universum

# Axioms of Probability

Kans van een gebeurtenis wordt genoteerd als P(A)

1. P(A) >= 0
2. P(Ω) = 1
3. Als gebeurtenissen A en B niet overlappen => Kans van A OF B = P(A) + P(B)

## Eigenschappen

Uit de axioma's kan je drie eigenschappen afleiden:

- Complement rule: **P(¬A) = 1 - P(A)** => de kans dat het niet A is, is 1 - de kans van A
- Onmogelijke gebeurtenis: **P(∅) = 0**
- General sum rule = **P(A V B) = P(A) + P(B) - P(A Λ B)** => de kans van A of B is de som van de kansen voor A en B min de kans van een situatie waarin A en B zich samen voordoen (= overlap)

# Gebeurtenissen


Independent events = Gebeurtenissen waarvan het voorkomen de kans van het voorkomen van een andere gebeurtenis niet beïnvloedt.

**P(A Λ B) = P(A)P(B)** => De kans dat A en B zich samen voordoen is de kans van A maal de kans van B

Als de gebeurtenissen niet aan deze formule voldoen, zijn ze dependent.

**P(B|A)** => lezen als: De kans dat B zich voordoet als A zich voorgedaan heeft (= **conditional probability**)

# Discrete random variable

= Stochastische waarde

-> Kwantiteit waarvan de waarde bepaald wordt door een willekeurige gebeurtenis.

Als er een kans vasthangt aan de variabelen, krijg je een probability distribution.

Een discrete random variabele X:
- neemt een waarde x<sub>i</sub> aan.
- heeft een probabiliteit dat X als waarde x<sub>i</sub> krijgt => P(X = x<sub>i</sub>) = p<sub>i</sub>
- p<sub>i</sub> is altijd 0 of hoger, en de som van alle probabiliteiten moet 1 zijn.

## Probability mass function

Geeft de probabiliteit dat een random variable een bepaalde waarde aanneemt.

Notatie: *f*x(waarde) <br>
Dit is hetzelfde als: P(X = waarde)

**Voorbeeld:**

Mapping van getrokken kaarten uit een kaartenspel. De speler verdient de gemapte waarde als hij deze kaart trekt:

|Kaart      |Gemapte waarde |
|-----------|---------------|
| boer      | 1             |
| dame      | 2             |
| heer      | 3             |
| anders    | 0             |

fx(3) = P(X = 3) = P(heer) = 4/52 = 1/13

fx(0) + fx(1) + fx(2) + fx(3) = 1 (alle waarden samen is altijd 1)

Je kan hiermee makkelijk verschillende probabiliteiten berekenen. <br> 
vb. Kans dat de score maximaal 1 is = fx(0) + fx(1)

# Continuous Random Variable

Variabele met ontelbaar veel mogelijke waarden.

-> Geen nut om voor de variabele X een exacte waarde te meten, wel werken met intervallen.

Frequency density diagrammen -> oppervlak van de bars / onder de curve is altijd gelijk aan 1. <br>
Als de intervallen kleiner worden, begin je willekeurige pieken te zien. Dit komt doordat het aantal metingen beperkt is -> in een volgend experiment zou je andere waarden krijgen.

**Variantie en verwachte waarde van continuous random variables moeten niet gekend zijn. Ze worden hetzelfde berekend als voor de discrete random variables, maar er worden integralen gebruikt.-> NIET TE KENNEN**

# Normale distributie

= Gaussian distribution


-> Continuous probability distribution met enkele speciale eigenschappen:
- symmetrisch t.o.v. de verwachte waarde (mean) => bell-curve
- oneindig in beide richtingen
- piek in het midden (op de mean)
- continue
- 95% van de waarden liggen binnen twee standaardafwijkingen van het gemiddelde
- 99% van de waarden liggen binnen drie standaardafwijkingen van het gemiddelde
- gebied onder de curve stelt 1 voor
- beschreven in termen van mean en standard deviation

Notatie: X ~ Nor(μ, σ)

## Python functions

Er bestaan functies binnen SciPy die gerelateerd zijn aan normale distributie.

## Parameters van een normale distributie

Alle normale distributies hebben dezelfde vorm.

- Mean: Als de mean lager is, ligt de piek meer naar links.
- Standaardafwijking: Hoe lager de afwijking, hoe hoger de piek.

## Standard normal distribution

Is een normale distributie met mean 0 en variantie/standaardafwijking 1.

Notatie: Z ~ N(0,1)
