1. Set up the correct crosstab.

```python
observed = pd.crosstab(df['Wijn'], df['Muziek'])
observed
```

2. Determine the marginal totals.

```python
# observed = pd.crosstab(df['Wijn'], df['Muziek'], margins=True)
# bovenstaande is mogelijk als je een nieuwe crosstab maakt
```

```python
observed['Total'] = observed.sum(axis=1)
observed.loc['Total'] = observed.sum()
observed
```

3. Determine the expected results.

```python
expected = np.outer(observed['Total'], observed.loc['Total']) / observed['Total'].sum()
expected
```

4. Calculate the χ² test statistic.

```python
H0 = 'There is insufficient evidence to reject that music and wine are independent variables.'
H1 = 'There is evidence that suggests music and wine are dependent variables'
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(observed)
```

```python
print(H0 if p_value > 0.05 else H1)
```

5. Calculate Cramér's V. What can you conclude from this?

```python
cramer_v = stats.contingency.association(observed, method='cramer')
print(f'Cramér\'s V = {cramer_v}, which indicates that there is a moderate association between wine played in the supermarket and music played.')
```
