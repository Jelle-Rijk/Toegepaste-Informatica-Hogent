## Exercise 1
Criminologists have long debated whether there is a relationship between weather conditions and the number of crimes committed. In 1988, an article “Is there a
Season for Homicide?” (Criminology, 1988: 287-296) classified 1361 homicides by season, resulting
in the data below.  
Is there an association between season and the number of murders committed?


```python
dfmurders = pd.DataFrame(data={
    'season': ['winter', 'spring','summer','autumn'],
    'number_of_murders': [328, 334, 372, 327]
})
```

```python
H0 = 'There is insufficient evidence to suspect that seasons influence the number of murders committed.'
H1 = 'There is enough evidence to conclude that seasons influence the number of murders committed.'

dfmurders['expected'] = dfmurders['number_of_murders'].sum() / len(dfmurders['season'])
dfmurders.head()
```

```python
alpha = 0.05
chi2, p_value = stats.chisquare(f_obs=dfmurders['number_of_murders'], f_exp=dfmurders['expected'])
print(f'Chi2 = {chi2}\np = {p_value}')
print(H0 if p_value > alpha else H1)
```

## Exercise 2
A random sample of individuals was taken, and each individual was classified according to their level of education and their sense of spirituality.
Is there an association between the level of education and the sense of spirituality?

```python
dfspirituality = pd.DataFrame(data={
    'sense_of_spirituality': ['lot', 'moderate','little','none'],
    'alfa_scientist': [56, 162, 198, 211],
    'beta_scientist': [56, 223, 243, 239],
    'not_scientist': [109, 164, 74, 28]
})
dfspirituality.head()
```

```python
dfspirituality = dfspirituality.set_index('sense_of_spirituality')
dfspirituality['total'] = dfspirituality.sum(axis=1)
dfspirituality.loc['total'] = dfspirituality.sum()
dfspirituality.head()
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(dfspirituality)
print(f'Chi2 = {chi2}\np_value = {p_value}')
alpha = 0.05
print("There is not enough evidence to say that spirituality and level of education are dependent" if p_value > alpha else "The sample suggests there is a level of dependence between spirituality and level of education")
```

## Exercise 4

In an experiment, 150 patients were divided into three groups of 45, 45 and 60 patients each. Two groups were given a new drug in different doses and the third group was given a placebo. The way the patients felt after using the drug are shown in the data frame below. Is there evidence at the 5% significance level that there is a difference in the condition between the three groups of patients?



```python
dfexperiment = pd.DataFrame(data={
    'state': ['better', 'equal', 'worse'],
    'placebo': [16, 20, 9],
    'half_dosis': [17, 18, 10],
    'full_dosis': [26, 20, 14]
}).set_index('state')
dfexperiment.head()
```

```python
dfexperiment['total'] = dfexperiment.sum(axis=1)
dfexperiment.loc['total'] = dfexperiment.sum()
dfexperiment.head()
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(dfexperiment)
print(f"Chi2 = {chi2}\np = {p_value}")
```

```python
alpha = 0.05
print("There is not enough evidence to suggest that the dosis and the patients felt after the experiment are dependent." if p_value > alpha else "There is sufficient evidence to suggest that the dosis had an effect on the way people felt")
```
