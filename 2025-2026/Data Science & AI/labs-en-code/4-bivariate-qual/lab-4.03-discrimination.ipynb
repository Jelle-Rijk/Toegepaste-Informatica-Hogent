```python
# Formuleer de hypotheses
H0 = 'There is not enough evidence to prove there is an association between the variables.'
H1 = 'There is an association between the variables.'

# Stel alpha vast
alpha = 0.05

# Bereken de teststatistiek = GOODNESS OF FIT TEST
types = ['Afro American', 'Others']
observed_values = np.array([15, 390])
expected_probabilities = np.array([0.057, 0.943])
expected = expected_probabilities * sum(observed_values)

print(expected)
```

```python
n = sum(observed_values)
k = len(observed_values)
dof = k - 1
chi2, p = stats.chisquare(f_exp=expected, f_obs=observed_values)

print(f'chi2={chi2}, p-value={p}')
print(H0 if p > alpha else H1)
```
