## Exercise 3 - Discrimination in schoolteacher hiring

African Americans in a St. Louis suburb sued the city 
claiming they were discriminated against in schoolteacher hiring. Of the city's population, 5.7% were 
African American; of 405 teachers in the school system, 15 were African American. Set up appropriate 
hypotheses and determine whether African Americans 
are underrepresented.  
Calculate the standardized residuals. 

Results of the main calculations:
- Chi-squared        χ² = 3.0027
- Critical value      g = 3.8415
- p-value             p = 0.0831
- standardized residuals for african american = -1.7328 > - 2

```python
H0 = "There is no reason to believe the observed data does not fit the population."
H1 = "There is enough evidence to believe that the sample does not represent the population well."

observed = [15, 405 - 15]
ratios = [0.057, 1 - 0.057]
expected = [x * 405 for x in ratios]
print(expected)
```

```python
chi2, p_value = stats.chisquare(f_obs=observed, f_exp=expected)
print(f"Chi2 = {chi2}")
print(f"p-value = {p_value}")
critical = stats.chi2.isf(0.05, 1)
print(f"Critical value = {critical}")

standardized_residual = (observed[0] - expected[0]) / np.sqrt(expected[0] * (1 - ratios[0]))
print(f"Standardized residual for african american: {standardized_residual}")
```
