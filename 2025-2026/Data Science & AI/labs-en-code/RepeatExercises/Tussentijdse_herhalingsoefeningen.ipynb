Een producent van sinaasappelsap koopt al zijn sinaasappels bij één grote sinaasappelproducent. De hoeveelheid sap die uit elk van deze sinaasappels wordt geperst is ongeveer normaal verdeeld, met een gemiddelde van 133 gram en een standaardafwijking van 11,3 gram.

1. Wat is de kans dat een willekeurig gekozen sinaasappel tussen de 133 gram en 142 gram sap bevat?
2. 77% van de sinaasappels zal minstens hoeveel gram sap bevatten?
3. 80% van de sinaasappels bevat een hoeveelheid sap die tussen welke twee waarden (symmetrisch verdeeld) rond het gemiddelde ligt?

```python
# Wat is de kans dat een willekeurig gekozen sinaasappel tussen de 133 gram en 142 gram sap bevat?
pop_mean = 133
pop_stdev = 11.3

kans_133_tot_142 = stats.norm.cdf(x=142, loc=pop_mean, scale=pop_stdev) - stats.norm.cdf(x=133, loc=pop_mean, scale=pop_stdev)
print(f'De kans dat een sinaasappel tussen de 133 en 142 gram sap bevat is {np.round(kans_133_tot_142 * 100, 2)}%')
```

```python
# 77% van de sinaasappels zal minstens hoeveel gram sap bevatten?
prob_x = (100 - 77) / 100
minstens = stats.norm.isf(q=1-prob_x, loc=pop_mean, scale=pop_stdev)
print(f'77% van de sinaasappels zal minstents {minstens} gram sap bevatten.')
```

```python
# 80% van de sinaasappels bevat een hoeveelheid sap die tussen welke twee waarden (symmetrisch verdeeld) rond het gemiddelde ligt?
min = stats.norm.isf(q=0.9, loc=pop_mean, scale=pop_stdev) 
max = stats.norm.isf(q=0.1, loc=pop_mean, scale=pop_stdev)
print(f"80% van de sinaasappels zal een hoeveelheid tussen de {np.round(min, 2)} en {np.round(max, 2)} gram bevatten.")
```

Een doos bevat twee rode en twee groene ballen. Een deelnemer aan een spelshow kiest een willekeurig balletje. Als de bal groen is, krijgt hij geen prijs voor de trekking en legt hij de bal aan de kant. Als de bal rood is, ontvangt hij $1000 en legt hij de bal terug in de doos. Het spel is afgelopen als beide groene ballen getrokken zijn of na drie trekkingen, wat het eerste komt.

1. Hoe groot is de kans dat de deelnemer helemaal geen prijs krijgt?
2. Hoe groot is de kans dat er 1 rode bal wordt getrokken?
3. Hoe groot is de kans dat er 2 rode ballen worden getrokken?
4. Hoe groot is de kans dat er 3 rode ballen worden getrokken?
5. Bereken uit 1. en 2. en 3. en 4. de verwachte prijs van de deelnemer.

```python
# 0 rode ballen = GG
p_geen_prijs = (1/2) * (1/3)

print(f'De kans dat de deelnemer geen prijs krijgt, is {p_geen_prijs}')

# 1 rode bal
# Mogelijkheden: RGG / GRG
p_1_rode_bal = (1/2 * 1/2 * 1/3) + (1/2 * 2/3 * 1/3)

print(f"De kans dat de deelnemer één rode bal trekt, is {p_1_rode_bal}")

# 2 rode ballen
# Mogelijkheden: RRG / RGR / GRR
p_2_rode_ballen = (1/2 * 1/2 * 1/2) + (1/2 * 1/2 * 2/3) + (1/2 * 2/3 * 2/3)
print(f"De kans dat de deelnemer twee rode ballen trekt, is {p_2_rode_ballen}")


# 3 rode ballen
# Mogelijkheden: RRR
p_3_rode_ballen = (1/2 * 1/2 * 1/2)
print(f"De kans dat de deelnemer drie rode ballen trekt, is {p_3_rode_ballen}")

# verwachte prijs
verwachte_prijs = 0 * p_geen_prijs + 1000 * p_1_rode_bal + 2000 * p_2_rode_ballen + 3000 * p_3_rode_ballen
print(f"De verwachte prijs is €{np.round(verwachte_prijs, 2)}")
```

Een artikel rapporteerde dat voor een steekproef van 50 keukens met gaskooktoestellen die gedurende een ganse week werden gemonitord, het gemiddelde CO2-niveau in de steekproef 654.16 ppm was en de standaardafwijking 164.43 ppm.

Bereken en interpreteer een 95% confidence interval voor het ware gemiddelde CO2 -niveau in de huizen die betrokken waren in de steekproef.

```python
alpha = 0.05
sample_mean = 654.16
sample_std = 164.43
n = 50

interval = stats.t.interval(1-alpha, loc=sample_mean, scale=sample_std/np.sqrt(n), df=n-1)
print(f"Er is een kans van {np.round((1 - alpha) * 100)}% dat het ware gemiddelde tussen {interval[0]} en {interval[1]} ligt.")
```

De vulhoeveelheid van flessen frisdrank is normaal verdeeld, met een gemiddelde van 2,0 liter en een standaardafwijking van 0,05 liter. Als flessen minder dan 95% van de vermelde netto-inhoud bevatten (in dit geval 1,90 liter), kan de fabrikant een boete krijgen . Flessen met een netto inhoud van meer dan 2,10 liter kunnen overmatig morsen bij het openen. Welk deel van de flessen bevat

1. tussen 1,90 en 2,0 liter?
2. minder dan 1,90 liter of meer dan 2,10 liter?
3. 99% van de flessen bevat minstens hoeveel frisdrank?

```python
pop_mean = 2
pop_stdev = 0.05
min_content = 1.9
max_content = 2.1

within_content_range = stats.norm.sf(1.9, loc=pop_mean, scale=pop_stdev) - stats.norm.sf(2, loc=pop_mean, scale=pop_stdev)
print(f"{np.round(within_content_range * 100, 2)}% van de flessen bevat tussen de {min_content}l en {2}l.")

teveel_teweinig = stats.norm.cdf(1.9, loc=pop_mean, scale=pop_stdev) + stats.norm.sf(2.1, loc=pop_mean, scale=pop_stdev)
print(f"{np.round(teveel_teweinig * 100, 2)}% van de flessen bevat te veel of te weinig frisdrank.")

minstens_99_procent = stats.norm.isf(0.99, loc=pop_mean, scale=pop_stdev)
print(f"99% van de flessen bevat minstens {np.round(minstens_99_procent,2)}l frisdrank.")
```
