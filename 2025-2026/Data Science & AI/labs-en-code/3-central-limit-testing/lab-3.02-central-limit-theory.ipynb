## Exercise 1 - Cholesterol 

A health survey between 1988 and 1994 indicated that the mean cholesterol value for women between the age of 20 and 29 was 183 mg/dl, with a standard deviation equal to 36. We now take a random sample of 81 women.

1. Plot the probability distribution of the sample mean $\overline{x}$.
2. What is the probability of the sample mean being $\overline{x} < 185$? (Answer ≈ 69,1%)
3. What is the probability of $175 < \overline{x} < 185$ (Answer ≈ 66,9%)
4. What is the probability of $\overline{x} > 190$ (Answer ≈ 4,0%)

```python
mean = 183
s = 36
n = 81

# 1. Plot the probability distribution of the sample mean.
x = np.linspace(mean - (s / 2), mean + (s / 2), num=201)
y = stats.norm.pdf(x, loc=mean, scale=s/np.sqrt(n))
plt.plot(x, y)

# 2. What is the probability of the sample mean being < 185? (Answer ≈ 69,1%)
lt_185 = stats.norm.cdf(185, loc=mean, scale=s/np.sqrt(n))
print(f"The probability of the sample mean being lower than 185 is {np.round(lt_185 * 100, 1)}%.")

# 3. What is the probability of 175 < sample mean < 185$ (Answer ≈ 66,9%)
between_175_185 = lt_185 - stats.norm.cdf(175, loc=mean, scale=s/np.sqrt(n))
print(f"The probability of the sample mean being between 175 and 185 is {np.round(between_175_185 * 100, 1)}%.")

# 4. What is the probability of sample mean > 190$ (Answer ≈ 4,0%)
gt_190 = stats.norm.sf(190, loc=mean, scale=s/np.sqrt(n))
print(f"The probability of the sample mean being bigger than 190 is {np.round(gt_190 * 100, 1)}%.")
```

## Exercise 2

A random sample of 64 is drawn from a population with unknown distribution. However, the expectation and standard deviation of the population are known: $\mu = 20$ and $\sigma = 16$.

1. Plot the probability distribution of the sample mean.
2. Calculate the z-score for $\overline{x_1} = 15.5$  
3. Calculate the z-score for $\overline{x_2} = 23$ 
4. Determine the probability that 16 < $\overline{x}$ < 22 (Answer ≈ 81.9%)

```python
mu = 20
sigma = 16
n=64

# 1. Plot the probability distribution of the sample mean.
x = np.linspace(mu - sigma, mu + sigma, num=201)
y = stats.norm.pdf(x, loc=mu, scale=sigma/np.sqrt(n))
plt.plot(x,y)

# 2. Calculate the z-score for $\overline{x_1} = 15.5$  
z = (15.5 - mu) / sigma
print(f"The z-score for 15.5 is {z}.")

# 3. Calculate the z-score for $\overline{x_2} = 23$ 
z = (23 - mu) / sigma
print(f"The z-score for 23 is {z}.")

# 4. Determine the probability that 16 < $\overline{x}$ < 22 (Answer ≈ 81.9%)
between_16_22 = stats.norm.cdf(22, loc=mu, scale=sigma/np.sqrt(n)) - stats.norm.cdf(16, loc=mu, scale=sigma/np.sqrt(n))
print(f"The probability that x is between 16 and 22 is {np.round(between_16_22 * 100, 1)}%.")
```
