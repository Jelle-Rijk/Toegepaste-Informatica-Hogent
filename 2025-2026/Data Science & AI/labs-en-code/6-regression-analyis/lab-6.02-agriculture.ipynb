3. Draw a scatter plot of the number of farms versus the year.


```python
sns.scatterplot(data=farms, x='year', y='number_of_farms')
```

4. Is there a positive / negative relationship between year and number_of_farms?

    $R = -0.9861066349492859$

```python
r = np.corrcoef(x=farms['year'], y=farms['number_of_farms'])[0][1]
print(f'R = {r}')
print(f'R indicates that there is a {'positive' if r > 0 else 'negative' if r < 0 else 'no'} relationship between year and number of farms.')
```

6. Is there a strong relationship between year and 'number_of_farms'?

    $R^2 = 0.9724062954910041$

```python
r2 = r**2
print(f'R2 = {r2}')
print('Since R2 is around 0.97 there is a very strong relationship between the year and number of farms. 97% of the variance in the number of farms can be explained by the year.')
```

7. Draw a scatter plot of the 'average_area_per_farm_(ha)' versus the year.

```python
sns.scatterplot(data=farms, x='year', y='average_area_per_farm_(ha)')
```

8. What will be the 'average_area_per_farm_(ha)' in 2035?

Average area in 2035 = 34.91987804878045

```python
avg_area = 'average_area_per_farm_(ha)'
b1, b0 = np.polyfit(x=farms['year'], y=farms[avg_area], deg=1)
avg_in_2035 = b0 + b1 * 2035
print(f'The average area in 2035 can be estimated to be {np.round(avg_in_2035, 4)} ha.')
```

8. Calculate the total agricultural area for each year.   
Calculate for each year the change of the total area relative to 1980.  
Make a plot.

image.png


```python
farms['total_area'] = farms['number_of_farms'] * farms[avg_area]
total_area_1980 = farms.loc[farms['year'] == 1980]['total_area'][0]
print(total_area_1980)

farms['relative_area'] = farms['total_area'].div(total_area_1980)

plot = sns.scatterplot(data=farms, x='year', y='relative_area')
plot.set_ylim((0.85, 1.05))
```
